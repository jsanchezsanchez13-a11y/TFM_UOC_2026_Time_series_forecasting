### Creacion de la columna de ventas reconstruida

```python
df_venta_stock = df_venta_stock.sort_values(["producto", "fecha"]).copy()
df_venta_stock["udsVenta_reconstruida"] = df_venta_stock["udsVenta"].astype(float)
df_venta_stock["dia_semana"] = df_venta_stock["fecha"].dt.dayofweek
```

si hay rotura de stock
busca días cercanos del mismo día de la semana
si no encuentra suficientes, usa días cercanos
si tampoco, usa la media histórica del producto

```python
def reconstruir_ventas(grupo):
    grupo = grupo.sort_values("fecha").copy()
    
    # Días válidos para estimar demanda
    ventas_validas = grupo.loc[
        (grupo["rotura_stock"] == 0) & (grupo["udsVenta"] > 0),
        "udsVenta"
    ]
    media_historica = ventas_validas.mean()
    
    # SOLO reconstruimos si hubo rotura y además ventas = 0
    for idx in grupo.index[
        (grupo["rotura_stock"] == 1) & (grupo["udsVenta"] == 0)
    ]:
        fecha_i = grupo.loc[idx, "fecha"]
        dow_i = grupo.loc[idx, "dia_semana"]
        
        # 1) mismo día de semana, ventana ±28 días
        candidatos_dow = grupo[
            (grupo["rotura_stock"] == 0) &
            (grupo["udsVenta"] > 0) &
            (grupo["dia_semana"] == dow_i) &
            (grupo["fecha"] >= fecha_i - pd.Timedelta(days=28)) &
            (grupo["fecha"] <= fecha_i + pd.Timedelta(days=28))
        ]["udsVenta"]
        
        if len(candidatos_dow) >= 2:
            grupo.loc[idx, "udsVenta_reconstruida"] = candidatos_dow.mean()
            continue
        
        # 2) días cercanos ±7 días
        candidatos_cercanos = grupo[
            (grupo["rotura_stock"] == 0) &
            (grupo["udsVenta"] > 0) &
            (grupo["fecha"] >= fecha_i - pd.Timedelta(days=7)) &
            (grupo["fecha"] <= fecha_i + pd.Timedelta(days=7))
        ]["udsVenta"]
        
        if len(candidatos_cercanos) >= 2:
            grupo.loc[idx, "udsVenta_reconstruida"] = candidatos_cercanos.mean()
            continue
        
        # 3) media histórica
        grupo.loc[idx, "udsVenta_reconstruida"] = media_historica
    
    return grupo
```

```python
df_venta_stock = (
    df_venta_stock
    .sort_values(["producto", "fecha"])
    .groupby("producto")
    .apply(reconstruir_ventas, include_groups=False)
    .reset_index()
    .drop(columns="level_1", errors="ignore")
)
```

```python
df_venta_stock
```

```python
df_venta_stock["udsVenta_reconstruida"] = df_venta_stock["udsVenta_reconstruida"].fillna(
    df_venta_stock["udsVenta"]
)
df_venta_stock["udsVenta_reconstruida"] = df_venta_stock["udsVenta_reconstruida"].round().astype(int)
```

```python
import plotly.graph_objects as go

productos = [1, 2, 3]

df_plot = df_venta_stock[df_venta_stock["producto"].isin(productos)].copy()
df_plot = df_plot.sort_values(["producto", "fecha"])

for prod in productos:
    df_p = df_plot[df_plot["producto"] == prod]
    
    fig = go.Figure()
    
    # Ventas reales
    fig.add_trace(go.Scatter(
        x=df_p["fecha"],
        y=df_p["udsVenta"],
        mode="lines",
        name="Venta real"
    ))
    
    # Ventas reconstruidas
    fig.add_trace(go.Scatter(
        x=df_p["fecha"],
        y=df_p["udsVenta_reconstruida"],
        mode="lines",
        line=dict(dash="dash"),
        name="Reconstruida"
    ))
    
    # Roturas de stock
    df_rotura = df_p[df_p["rotura_stock"] == 1]
    
    fig.add_trace(go.Scatter(
        x=df_rotura["fecha"],
        y=df_rotura["udsVenta"],
        mode="markers",
        marker=dict(size=10, color="red", symbol="x"),
        name="Rotura de stock"
    ))
    
    fig.update_layout(
        title=f"Producto {prod}",
        xaxis_title="Fecha",
        yaxis_title="Unidades",
        template="plotly_white"
    )
    
    # ✅ Guardar imagen con nombre dinámico
    fig.write_image(f"imgs/producto_{prod}_venta_real_vs_reconstruida.png", scale=2)
    
    fig.show()
```

```python
productos = [1, 2, 3]

df_plot = df_venta_stock[df_venta_stock["producto"].isin(productos)].copy()
df_plot["fecha"] = pd.to_datetime(df_plot["fecha"])
df_plot = df_plot.sort_values(["producto", "fecha"])

for prod in productos:
    df_p = df_plot[df_plot["producto"] == prod].copy()
    
    roturas_2025 = df_p[
        (df_p["rotura_stock"] == 1) &
        (df_p["fecha"] >= "2025-01-01") &
        (df_p["fecha"] <= "2025-12-31")
    ]["fecha"]
    
    for fecha_rotura in roturas_2025:
        inicio = fecha_rotura - pd.Timedelta(days=30)
        fin = fecha_rotura + pd.Timedelta(days=30)
        
        df_win = df_p[(df_p["fecha"] >= inicio) & (df_p["fecha"] <= fin)]
        df_rotura = df_win[df_win["rotura_stock"] == 1]
        
        fig = go.Figure()
        
        fig.add_trace(go.Scatter(
            x=df_win["fecha"],
            y=df_win["udsVenta"],
            mode="lines+markers",
            name="Venta real"
        ))
        
        fig.add_trace(go.Scatter(
            x=df_win["fecha"],
            y=df_win["udsVenta_reconstruida"],
            mode="lines+markers",
            line=dict(dash="dash"),
            name="Reconstruida"
        ))
        
        fig.add_trace(go.Scatter(
            x=df_rotura["fecha"],
            y=df_rotura["udsVenta"],
            mode="markers",
            marker=dict(size=10, color="red", symbol="x"),
            name="Rotura de stock"
        ))
        
        fig.update_layout(
            title=f"Producto {prod} - Ventana alrededor de {fecha_rotura.date()}",
            xaxis_title="Fecha",
            yaxis_title="Unidades",
            template="plotly_white"
        )
        # Guardar imagen
        fecha_str = fecha_rotura.strftime("%Y-%m-%d")
        fig.write_image(f"imgs/producto_{prod}_rotura_{fecha_str}.png", scale=2)
        
        fig.show()
```

## Venta medias por producto producidas en 2025 con promociones

```python
productos = [1, 2, 3]

df_plot = df_venta_calendario_promociones[
    df_venta_calendario_promociones["producto"].isin(productos)
].copy()

df_plot["fecha"] = pd.to_datetime(df_plot["fecha"])
df_plot = df_plot.sort_values(["producto", "fecha"])

for prod in productos:
    df_p = df_plot[
        (df_plot["producto"] == prod) &
        (df_plot["fecha"].between("2025-01-01", "2025-12-31"))
    ].copy()
    
    fig = go.Figure()
    
    # Venta real
    fig.add_trace(go.Scatter(
        x=df_p["fecha"],
        y=df_p["udsVenta"],
        mode="lines",
        name="Venta real"
    ))
    
    # Reconstruida
    fig.add_trace(go.Scatter(
        x=df_p["fecha"],
        y=df_p["udsVenta_reconstruida"],
        mode="lines",
        line=dict(dash="dash"),
        name="Reconstruida"
    ))
    
    # Promociones (puntos)
    df_promo = df_p[df_p["promociones"] == 1]
    
    fig.add_trace(go.Scatter(
        x=df_promo["fecha"],
        y=df_promo["udsVenta_reconstruida"],
        mode="markers",
        marker=dict(size=8, color="green", symbol="diamond"),
        name="Promoción"
    ))
    
    # sombreado vertical para promos
    for fecha in df_promo["fecha"]:
        fig.add_vline(x=fecha, line_width=1, opacity=0.1)
    
    fig.update_layout(
        title=f"Producto {prod} - Ventas y promociones (2025)",
        xaxis_title="Fecha",
        yaxis_title="Unidades",
        template="plotly_white"
    )
            # Guardar imagen
    fig.write_image(f"imgs/producto_{prod}_rotura_promocion.png", scale=2)
    fig.show()
```

```python
# Filtrar datos
ventas_en_promocion = df_venta_calendario_promociones[
    df_venta_calendario_promociones['promociones'] == 1
]

ventas_fuera_de_promocion = df_venta_calendario_promociones[
    df_venta_calendario_promociones['promociones'] == 0
]

# Agregación diaria (IMPORTANTE: sobre cada subset)
sumatorio_ventas_en_promocion = (
    ventas_en_promocion
    .groupby('fecha')['udsVenta_reconstruida']
    .sum()
)

sumatorio_ventas_fuera_de_promocion = (
    ventas_fuera_de_promocion
    .groupby('fecha')['udsVenta_reconstruida']
    .sum()
)

# Gráfico
plt.figure(figsize=(12,6))

plt.plot(
    sumatorio_ventas_en_promocion.index,
    sumatorio_ventas_en_promocion.values,
    label='Ventas en promoción',
    alpha=0.7
)

plt.plot(
    sumatorio_ventas_fuera_de_promocion.index,
    sumatorio_ventas_fuera_de_promocion.values,
    label='Ventas fuera de promoción',
    alpha=0.7,
    linestyle='--'
)

# Etiquetas
plt.title('Evolución de ventas: promoción vs no promoción', fontsize=16)
plt.xlabel('Fecha', fontsize=12)
plt.ylabel('Unidades vendidas', fontsize=12)

plt.legend()
plt.xticks(rotation=45)
plt.grid(alpha=0.3)
plt.tight_layout()

# Guardar imagen
plt.savefig("imgs/ventas_promocion_vs_no_promocion.png", dpi=300)

plt.show()
```

## Descomposion de la serie temporal

```python
from statsmodels.tsa.seasonal import STL

# Descomposición de la serie temporal
productos = [1]

for prod in productos:
    df_p = df_venta_calendario_promociones[
        df_venta_calendario_promociones["producto"] == prod
    ].copy()
    
    df_p = df_p.set_index("fecha").sort_index()
    
    stl = STL(df_p["udsVenta_reconstruida"], period=30)
    result = stl.fit()
    
    fig = result.plot()
    fig.set_size_inches(12, 8)
    plt.title(f"Producto {prod}")
    plt.show()
```
