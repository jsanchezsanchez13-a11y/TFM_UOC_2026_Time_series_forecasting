```python
import os
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.dates as mdates

# Crear carpeta si no existe
os.makedirs("imgs", exist_ok=True)

# Asegurar datetime
df_udVenta['fecha'] = pd.to_datetime(df_udVenta['fecha'])

# Dataset test
val_end = pd.to_datetime("2025-12-31")
df_test = df_udVenta[df_udVenta['fecha'] > val_end]

# Productos y clusters
productos = {
    1: 1,
    28: 0,
    228: 2
}

# Crear un plot individual por producto
for producto, cluster in productos.items():

    # Filtrar producto
    test_p = (
        df_test[df_test['producto'] == producto]
        .sort_values('fecha')
    )

    # Figura
    fig, ax = plt.subplots(figsize=(16, 6))

    # Serie real
    ax.plot(
        test_p['fecha'],
        test_p['udsVenta'],
        label='Real',
        linewidth=2
    )

    # Predicción
    ax.plot(
        test_p['fecha'],
        test_p['pred_lightgbm_round'],
        '--',
        linewidth=2,
        label='Predicción LightGBM'
    )

    # Título
    ax.set_title(
        f'Producto {producto} - Cluster {cluster}',
        fontsize=14
    )

    ax.set_xlabel('Fecha')
    ax.set_ylabel('Unidades')
    ax.grid(True)
    ax.legend()

    # Formato fechas
    ax.xaxis.set_major_locator(mdates.WeekdayLocator(interval=2))
    ax.xaxis.set_major_formatter(mdates.DateFormatter('%Y-%m-%d'))

    fig.autofmt_xdate()

    plt.tight_layout()

    # Guardar imagen
    plt.savefig(
        f"imgs/producto_{producto}_cluster_{cluster}.png",
        dpi=300,
        bbox_inches='tight'
    )

    plt.close()

print("Imágenes guardadas en /imgs")
```
